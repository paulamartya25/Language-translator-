```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import math
from typing import Optional,Tuple

class FFTBlk(nn.Module):
    def __init__(self,d=256,h=2,k=9,p=0.1):
        super().__init__()
        self.dh=d//h; self.h=h; self.n1=nn.LayerNorm(d)
        self.q=nn.Linear(d,d); self.k2=nn.Linear(d,d); self.v=nn.Linear(d,d)
        self.ao=nn.Linear(d,d); self.ad=nn.Dropout(p); self.sc=self.dh**-0.5
        self.n2=nn.LayerNorm(d); pad=(k-1)//2
        self.c1=nn.Conv1d(d,d*4,k,padding=pad); self.c2=nn.Conv1d(d*4,d,k,padding=pad); self.fd=nn.Dropout(p)
    def forward(self,x,mask=None):
        B,T,D=x.shape
        def sp(t): return t.view(B,-1,self.h,self.dh).transpose(1,2)
        xn=self.n1(x); Q,K,V=sp(self.q(xn)),sp(self.k2(xn)),sp(self.v(xn))
        s=(Q@K.transpose(-2,-1))*self.sc
        if mask is not None: s=s.masked_fill(mask.unsqueeze(1).unsqueeze(2),float('-inf'))
        sa=(self.ad(F.softmax(s,-1))@V).transpose(1,2).reshape(B,T,D)
        x=x+self.ad(self.ao(sa))
        xn=self.n2(x).transpose(1,2)
        x=x+self.fd(self.c2(self.fd(F.relu(self.c1(xn))))).transpose(1,2)
        return x

class VarPred(nn.Module):
    def __init__(self,d=256,h=256,k=3,p=0.1):
        super().__init__()
        pad=(k-1)//2
        self.c1=nn.Conv1d(d,h,k,padding=pad); self.c2=nn.Conv1d(h,h,k,padding=pad)
        self.n1=nn.LayerNorm(h); self.n2=nn.LayerNorm(h)
        self.lin=nn.Linear(h,1); self.drop=nn.Dropout(p)
    def forward(self,x):
        h=x.transpose(1,2)
        h=self.drop(F.relu(self.c1(h))); h=self.n1(h.transpose(1,2)).transpose(1,2)
        h=self.drop(F.relu(self.c2(h))); h=self.n2(h.transpose(1,2))
        return self.lin(h).squeeze(-1)

class LenReg(nn.Module):
    def forward(self,x,d,ml=None):
        B,T,D=x.shape; out=[]
        for i in range(B):
            dur=d[i].long().clamp(min=0); fr=x[i].repeat_interleave(dur,0); out.append(fr)
        if ml is None: ml=max(e.size(0) for e in out)
        res=torch.zeros(B,ml,D,device=x.device,dtype=x.dtype)
        for i,e in enumerate(out):
            l=min(e.size(0),ml); res[i,:l]=e[:l]
        return res,ml

class FastSpeech2TTS(nn.Module):
    def __init__(self,vocab_size=500,n_mels=80,d_model=256,n_enc=4,n_dec=4,n_heads=2,kernel_size=9,n_bins=256,dropout=0.1,max_len=1000):
        super().__init__()
        self.em=nn.Embedding(vocab_size,d_model,padding_idx=0)
        self.pe=nn.Embedding(max_len,d_model); self.drop=nn.Dropout(dropout)
        self.enc=nn.ModuleList([FFTBlk(d_model,n_heads,kernel_size,dropout) for _ in range(n_enc)])
        self.dp=VarPred(d_model); self.pp=VarPred(d_model); self.ep=VarPred(d_model)
        self.lr=LenReg()
        self.pit_em=nn.Embedding(n_bins+1,d_model); self.eng_em=nn.Embedding(n_bins+1,d_model); self.nb=n_bins
        self.dec=nn.ModuleList([FFTBlk(d_model,n_heads,kernel_size,dropout) for _ in range(n_dec)])
        self.ml=nn.Linear(d_model,n_mels); self.nm=n_mels; self.mxl=max_len
        k=5; pad=(k-1)//2; h2=512; layers=[]
        for i in range(5):
            ic=n_mels if i==0 else h2; oc=n_mels if i==4 else h2
            layers+=[nn.Conv1d(ic,oc,k,padding=pad),nn.BatchNorm1d(oc)]
            if i<4: layers.append(nn.Tanh())
        self.pn=nn.Sequential(*layers); self.pnd=nn.Dropout(0.5)
    def _q(self,v,nb):
        mn,mx=v.min(),v.max()
        if mx==mn: return torch.zeros_like(v,dtype=torch.long)
        bins=torch.linspace(mn.item(),mx.item(),nb+1,device=v.device)
        return torch.bucketize(v,bins[1:-1])
    def forward(self,ph,durs=None,pitch=None,energy=None):
        B,T=ph.shape; pos=torch.arange(T,device=ph.device).unsqueeze(0)
        x=self.drop(self.em(ph)+self.pe(pos.clamp(max=self.mxl-1)))
        for b in self.enc: x=b(x)
        dp=self.dp(x); pp=self.pp(x); ep=self.ep(x)
        dur_r=(durs if durs is not None else dp.detach().exp().round()).clamp(min=0)
        x,ml=self.lr(x,dur_r); B2,T2,D=x.shape
        pb=self._q(pp.detach() if pitch is None else pitch,self.nb)
        eb=self._q(ep.detach() if energy is None else energy,self.nb)
        def pb2(b,tgt):
            if b.size(1)>tgt: return b[:,:tgt]
            return F.pad(b,(0,tgt-b.size(1)))
        pb=pb2(pb,T2); eb=pb2(eb,T2)
        pos2=torch.arange(T2,device=x.device).unsqueeze(0)
        x=x+self.drop(self.pe(pos2.clamp(max=self.mxl-1)))+self.pit_em(pb)+self.eng_em(eb)
        for b in self.dec: x=b(x)
        mb=self.ml(x); ma=mb+self.pnd(self.pn(mb.transpose(1,2)).transpose(1,2))
        return mb,ma,dp,pp,ep
    def compute_loss(self,ph,mel_tgt,durs,pitch,energy):
        mb,ma,dp,pp,ep=self(ph,durs,pitch,energy)
        T=min(mb.size(1),mel_tgt.size(1))
        mel_loss=F.l1_loss(mb[:,:T],mel_tgt[:,:T])+F.l1_loss(ma[:,:T],mel_tgt[:,:T])
        T2=min(dp.size(1),durs.size(1))
        dur_loss=F.mse_loss(dp[:,:T2],durs.float()[:,:T2])
        return mel_loss+dur_loss

m=FastSpeech2TTS(vocab_size=500)
total=sum(p.numel() for p in m.parameters())
print(f'FastSpeech2TTS: {total:,} ({total/1e6:.1f}M) parameters')
del m; print('Architecture OK!')
```

```python
from datasets import load_dataset
from torch.utils.data import Dataset,DataLoader
import torchaudio,torch

SR=22050; N_MELS=80; N_FFT=1024; HOP=256
mel_t=torchaudio.transforms.MelSpectrogram(sample_rate=SR,n_fft=N_FFT,hop_length=HOP,n_mels=N_MELS)
amp_db=torchaudio.transforms.AmplitudeToDB(top_db=80)

CHARS=list(" abcdefghijklmnopqrstuvwxyz'.,!?")+['<pad>','<unk>']
char2id={c:i for i,c in enumerate(CHARS)}
VOCAB_SIZE=len(CHARS)
print(f'Char vocab: {VOCAB_SIZE}')

def txt2ids(t,ml=200):
    t=t.lower(); ids=[char2id.get(c,char2id['<unk>']) for c in t[:ml]]
    ids=ids+[char2id['<pad>']]*max(0,ml-len(ids))
    return torch.tensor(ids[:ml],dtype=torch.long)

class LJSDataset(Dataset):
    def __init__(self,hf_ds,max_mel=800,max_text=200):
        self.data=hf_ds; self.mm=max_mel; self.mt=max_text
    def __len__(self): return len(self.data)
    def __getitem__(self,idx):
        item=self.data[idx]
        audio=torch.tensor(item['audio']['array'],dtype=torch.float32).unsqueeze(0)
        if item['audio']['sampling_rate']!=SR:
            audio=torchaudio.functional.resample(audio,item['audio']['sampling_rate'],SR)
        mel=amp_db(mel_t(audio)).squeeze(0).T
        mel=mel[:self.mm]
        T,M=mel.shape
        if T<self.mm: mel=torch.cat([mel,torch.full((self.mm-T,M),mel.min())],0)
        text=item.get('text',item.get('normalized_text',''))
        ph=txt2ids(text,self.mt)
        n_ph=(ph!=char2id['<pad>']).sum().item()
        dph=self.mm//max(n_ph,1)
        durs=torch.full((self.mt,),dph,dtype=torch.long)
        pitch=mel.mean(1)[:self.mt]; energy=mel.abs().mean(1)[:self.mt]
        if pitch.size(0)<self.mt:
            pitch=torch.nn.functional.pad(pitch,(0,self.mt-pitch.size(0)))
            energy=torch.nn.functional.pad(energy,(0,self.mt-energy.size(0)))
        return ph,mel,durs,pitch,energy

try:
    lj=load_dataset('keithito/lj_speech',split='train',trust_remote_code=True)
    tts_ds=LJSDataset(lj)
except:
    class SynDS(Dataset):
        def __len__(self): return 2000
        def __getitem__(self,i):
            return (torch.randint(0,VOCAB_SIZE,(200,)),torch.randn(800,N_MELS),
                    torch.ones(200,dtype=torch.long)*4,torch.randn(200),torch.abs(torch.randn(200)))
    tts_ds=SynDS(); print('Using synthetic dataset')

train_loader=DataLoader(tts_ds,batch_size=16,shuffle=True,num_workers=2,pin_memory=True)
print(f'{len(tts_ds)} samples | {len(train_loader)} batches/epoch')
```

```python
import time,json
DEVICE=torch.device('cuda' if torch.cuda.is_available() else 'cpu')
NUM_EPOCHS=30; LR=1e-4
SAVE_PATH='/kaggle/working/tts_model.pt'

model=FastSpeech2TTS(vocab_size=VOCAB_SIZE,n_mels=N_MELS).to(DEVICE)
optimizer=torch.optim.AdamW(model.parameters(),lr=LR,weight_decay=1e-5)
scheduler=torch.optim.lr_scheduler.CosineAnnealingLR(optimizer,T_max=NUM_EPOCHS)
scaler=torch.cuda.amp.GradScaler(enabled=torch.cuda.is_available())
print(f'{sum(p.numel() for p in model.parameters())/1e6:.1f}M params | {DEVICE}')

best=float('inf'); hist=[]
for epoch in range(1,NUM_EPOCHS+1):
    model.train(); el=0.0; t0=time.time()
    for i,batch in enumerate(train_loader):
        ph,mel,durs,pit,eng=[b.to(DEVICE) for b in batch]; optimizer.zero_grad()
        with torch.cuda.amp.autocast(enabled=torch.cuda.is_available()):
            loss=model.compute_loss(ph,mel,durs,pit,eng)
        if torch.isnan(loss): continue
        scaler.scale(loss).backward(); scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(),1.0)
        scaler.step(optimizer); scaler.update(); el+=loss.item()
        if i%50==0: print(f'  Ep{epoch:02d} [{i}/{len(train_loader)}] loss={loss.item():.4f}')
    scheduler.step(); avg=el/max(1,len(train_loader))
    hist.append({'epoch':epoch,'loss':avg})
    print(f'Epoch {epoch:02d} avg={avg:.4f} | {time.time()-t0:.0f}s')
    torch.save({'epoch':epoch,'state':model.state_dict(),'loss':avg},SAVE_PATH.replace('.pt','_ckpt.pt'))
    if avg<best: best=avg; torch.save(model.state_dict(),SAVE_PATH); print(f'  Best! {best:.4f}')

with open('/kaggle/working/tts_history.json','w') as f: json.dump(hist,f)
print(f'Done! Best={best:.4f} | {SAVE_PATH}')
```
