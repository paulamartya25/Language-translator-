```python
# ==== ALL 3 CUSTOM ARCHITECTURES (inline) ====
import torch,torch.nn as nn,torch.nn.functional as F,torchaudio,math
from typing import Optional

# ---- 1. MicroConformerASR ----
class LogMelSpec(nn.Module):
    def __init__(self,sr=16000,n_fft=512,hop=160,n_mels=80):
        super().__init__()
        self.ms=torchaudio.transforms.MelSpectrogram(sample_rate=sr,n_fft=n_fft,hop_length=hop,n_mels=n_mels)
        self.db=torchaudio.transforms.AmplitudeToDB(top_db=80)
    def forward(self,w):
        m=self.db(self.ms(w)); return (m-m.mean(-1,keepdim=True))/(m.std(-1,keepdim=True)+1e-6)
class Swish(nn.Module):
    def forward(self,x): return x*torch.sigmoid(x)
class ConvSub(nn.Module):
    def __init__(self,ic,d):
        super().__init__()
        self.c=nn.Sequential(nn.Conv2d(1,256,3,2,1),nn.ReLU(),nn.Conv2d(256,256,3,2,1),nn.ReLU())
        self.p=nn.Linear(256*((ic+1)//4),d)
    def forward(self,x):
        x=self.c(x.unsqueeze(1)); B,C,F,T=x.size(); return self.p(x.permute(0,3,1,2).reshape(B,T,C*F))
class ConvMod(nn.Module):
    def __init__(self,d,k=31):
        super().__init__()
        self.n=nn.LayerNorm(d); self.p1=nn.Conv1d(d,2*d,1); self.glu=nn.GLU(1)
        self.dw=nn.Conv1d(d,d,k,padding=(k-1)//2,groups=d); self.bn=nn.BatchNorm1d(d)
        self.sw=Swish(); self.p2=nn.Conv1d(d,d,1); self.drop=nn.Dropout(0.1)
    def forward(self,x):
        r=x; x=self.n(x).transpose(1,2); x=self.sw(self.bn(self.dw(self.glu(self.p1(x)))))
        return self.drop(self.p2(x)).transpose(1,2)+r
class FFMod(nn.Module):
    def __init__(self,d,e=4,p=0.1):
        super().__init__()
        self.n=nn.LayerNorm(d); self.f1=nn.Linear(d,d*e); self.a=Swish()
        self.d1=nn.Dropout(p); self.f2=nn.Linear(d*e,d); self.d2=nn.Dropout(p)
    def forward(self,x): return x+0.5*self.d2(self.f2(self.d1(self.a(self.f1(self.n(x))))))
class MHSA(nn.Module):
    def __init__(self,d,h,p=0.1):
        super().__init__()
        self.dh=d//h; self.h=h; self.n=nn.LayerNorm(d)
        self.q=nn.Linear(d,d); self.k=nn.Linear(d,d); self.v=nn.Linear(d,d)
        self.o=nn.Linear(d,d); self.drop=nn.Dropout(p); self.sc=self.dh**-0.5
    def forward(self,x,mask=None):
        r=x; x=self.n(x); B,T,_=x.shape
        def sp(t): return t.view(B,-1,self.h,self.dh).transpose(1,2)
        Q,K,V=sp(self.q(x)),sp(self.k(x)),sp(self.v(x)); s=(Q@K.transpose(-2,-1))*self.sc
        if mask is not None: s=s.masked_fill(mask.unsqueeze(1).unsqueeze(2),float('-inf'))
        return self.drop(self.o((self.drop(F.softmax(s,-1))@V).transpose(1,2).reshape(B,T,-1)))+r
class ConfBlk(nn.Module):
    def __init__(self,d=256,h=4,k=31,p=0.1):
        super().__init__()
        self.ff1=FFMod(d,p=p); self.sa=MHSA(d,h,p); self.cv=ConvMod(d,k); self.ff2=FFMod(d,p=p); self.n=nn.LayerNorm(d)
    def forward(self,x,mask=None): return self.n(self.ff2(self.cv(self.sa(self.ff1(x),mask))))
class MHA(nn.Module):
    def __init__(self,d,h,p=0.1):
        super().__init__()
        self.dh=d//h; self.h=h; self.q=nn.Linear(d,d); self.k=nn.Linear(d,d); self.v=nn.Linear(d,d)
        self.o=nn.Linear(d,d); self.drop=nn.Dropout(p); self.sc=self.dh**-0.5
    def forward(self,q,k,v,mask=None):
        B=q.size(0)
        def sp(t): return t.view(B,-1,self.h,self.dh).transpose(1,2)
        Q,K,V=sp(self.q(q)),sp(self.k(k)),sp(self.v(v)); s=(Q@K.transpose(-2,-1))*self.sc
        if mask is not None: s=s.masked_fill(mask,float('-inf'))
        return self.o((self.drop(F.softmax(s,-1))@V).transpose(1,2).reshape(B,-1,self.h*self.dh))
class DecBlk(nn.Module):
    def __init__(self,d=256,h=4,f=1024,p=0.1):
        super().__init__()
        self.n1=nn.LayerNorm(d); self.sa=MHA(d,h,p); self.n2=nn.LayerNorm(d); self.ca=MHA(d,h,p)
        self.n3=nn.LayerNorm(d); self.ff=nn.Sequential(nn.Linear(d,f),nn.ReLU(),nn.Dropout(p),nn.Linear(f,d),nn.Dropout(p))
    def forward(self,t,m,tm=None,mm=None):
        x=t; x=x+self.sa(self.n1(x),self.n1(x),self.n1(x),tm)
        x=x+self.ca(self.n2(x),m,m,mm); return x+self.ff(self.n3(x))
class TDec(nn.Module):
    def __init__(self,vs,d=256,L=6,h=4,f=1024,p=0.1,ml=512):
        super().__init__()
        self.em=nn.Embedding(vs,d,padding_idx=0); self.pe=nn.Embedding(ml,d)
        self.drop=nn.Dropout(p); self.blocks=nn.ModuleList([DecBlk(d,h,f,p) for _ in range(L)]); self.n=nn.LayerNorm(d)
    def forward(self,t,m,tm=None,mm=None):
        B,T=t.size(); pos=torch.arange(T,device=t.device).unsqueeze(0)
        x=self.drop(self.em(t)+self.pe(pos))
        for b in self.blocks: x=b(x,m,tm,mm)
        return self.n(x)
class MicroConformerASR(nn.Module):
    def __init__(self,vocab_size=5000,n_mels=80,d_model=256,enc_layers=6,dec_layers=6,n_heads=4,kernel_size=31,ffn_dim=1024,dropout=0.1,max_len=512,ctc_weight=0.3):
        super().__init__()
        self.ctc_w=ctc_weight; self.fe=LogMelSpec(n_mels=n_mels); self.sub=ConvSub(n_mels,d_model)
        self.drop=nn.Dropout(dropout)
        self.enc=nn.ModuleList([ConfBlk(d_model,n_heads,kernel_size,dropout) for _ in range(enc_layers)])
        self.enc_n=nn.LayerNorm(d_model); self.dec=TDec(vocab_size,d_model,dec_layers,n_heads,ffn_dim,dropout,max_len)
        self.ctc=nn.Linear(d_model,vocab_size); self.lm=nn.Linear(d_model,vocab_size); self.lm.weight=self.dec.em.weight
    def encode(self,w,mask=None):
        x=self.sub(self.fe(w))
        for b in self.enc: x=b(x,mask)
        return self.enc_n(x)
    @torch.no_grad()
    def greedy_decode(self,w,max_len=200,bos=1,eos=2):
        self.eval(); m=self.encode(w); B=w.size(0)
        toks=torch.full((B,1),bos,dtype=torch.long,device=w.device)
        for _ in range(max_len):
            n=self.lm(self.dec(toks,m))[:,-1,:].argmax(-1,keepdim=True)
            toks=torch.cat([toks,n],1)
            if (n==eos).all(): break
        return toks

# ---- 2. MultilingualSeq2Seq ----
class RoPE(nn.Module):
    def __init__(self,dh,ml=2048):
        super().__init__(); inv=1.0/(10000**(torch.arange(0,dh,2).float()/dh)); self.register_buffer('inv',inv)
    def _rh(self,x): x1,x2=x[...,::2],x[...,1::2]; return torch.stack([-x2,x1],-1).flatten(-2)
    def forward(self,x):
        B,H,T,D=x.shape; t=torch.arange(T,device=x.device).float()
        f=torch.einsum('i,j->ij',t,self.inv); e=torch.cat([f,f],-1).unsqueeze(0).unsqueeze(0)
        return x*e.cos()+self._rh(x)*e.sin()
class MHARoPE(nn.Module):
    def __init__(self,d,h,p=0.1,rp=True):
        super().__init__()
        self.dh=d//h; self.h=h; self.rp_on=rp
        self.q=nn.Linear(d,d); self.k=nn.Linear(d,d); self.v=nn.Linear(d,d)
        self.o=nn.Linear(d,d); self.drop=nn.Dropout(p); self.sc=self.dh**-0.5
        if rp: self.rope=RoPE(self.dh)
    def forward(self,q,k,v,mask=None):
        B=q.size(0)
        def sp(t): return t.view(B,-1,self.h,self.dh).transpose(1,2)
        Q,K,V=sp(self.q(q)),sp(self.k(k)),sp(self.v(v))
        if self.rp_on: Q=self.rope(Q); K=self.rope(K)
        s=(Q@K.transpose(-2,-1))*self.sc
        if mask is not None: s=s.masked_fill(mask,float('-inf'))
        return self.drop(self.o((self.drop(F.softmax(s,-1))@V).transpose(1,2).reshape(B,-1,self.h*self.dh)))
class PnFFN(nn.Module):
    def __init__(self,d,f,p=0.1):
        super().__init__()
        self.n=nn.LayerNorm(d); self.f1=nn.Linear(d,f); self.f2=nn.Linear(f,d); self.drop=nn.Dropout(p); self.act=nn.GELU()
    def forward(self,x): return x+self.drop(self.f2(self.drop(self.act(self.f1(self.n(x))))))
class EncBlk2(nn.Module):
    def __init__(self,d=512,h=8,f=2048,p=0.1):
        super().__init__()
        self.n=nn.LayerNorm(d); self.sa=MHARoPE(d,h,p,True); self.ffn=PnFFN(d,f,p)
    def forward(self,x,mask=None): xn=self.n(x); return self.ffn(x+self.sa(xn,xn,xn,mask))
class DecBlk2(nn.Module):
    def __init__(self,d=512,h=8,f=2048,p=0.1):
        super().__init__()
        self.n1=nn.LayerNorm(d); self.sa=MHARoPE(d,h,p,True)
        self.n2=nn.LayerNorm(d); self.ca=MHARoPE(d,h,p,False); self.ffn=PnFFN(d,f,p)
    def forward(self,t,m,tm=None,mm=None):
        xn=self.n1(t); t=t+self.sa(xn,xn,xn,tm); xn=self.n2(t); t=t+self.ca(xn,m,m,mm); return self.ffn(t)
class MultilingualSeq2Seq(nn.Module):
    def __init__(self,vocab_size=32000,d_model=512,n_enc=6,n_dec=6,n_heads=8,ffn_dim=2048,dropout=0.1,max_len=512,pad_id=0):
        super().__init__()
        self.pad=pad_id; self.d=d_model; self.em=nn.Embedding(vocab_size,d_model,padding_idx=pad_id)
        self.enc=nn.ModuleList([EncBlk2(d_model,n_heads,ffn_dim,dropout) for _ in range(n_enc)])
        self.dec=nn.ModuleList([DecBlk2(d_model,n_heads,ffn_dim,dropout) for _ in range(n_dec)])
        self.en=nn.LayerNorm(d_model); self.dn=nn.LayerNorm(d_model)
        self.lm=nn.Linear(d_model,vocab_size,bias=False); self.lm.weight=self.em.weight; self.drop=nn.Dropout(dropout)
    def _cm(self,T,dev): return torch.triu(torch.ones(T,T,device=dev),1).bool().unsqueeze(0).unsqueeze(0)
    def encode(self,src):
        mask=(src==self.pad).unsqueeze(1).unsqueeze(2); x=self.drop(self.em(src))*math.sqrt(self.d)
        for b in self.enc: x=b(x,mask)
        return self.en(x)
    def decode(self,tgt,mem):
        T=tgt.size(1); tm=self._cm(T,tgt.device); x=self.drop(self.em(tgt))*math.sqrt(self.d)
        for b in self.dec: x=b(x,mem,tm)
        return self.dn(x)
    @torch.no_grad()
    def translate(self,src,max_len=200,bos=1,eos=2):
        self.eval(); m=self.encode(src); B=src.size(0)
        t=torch.full((B,1),bos,dtype=torch.long,device=src.device)
        for _ in range(max_len):
            h=self.decode(t,m); n=self.lm(h[:,-1,:]).argmax(-1,keepdim=True)
            t=torch.cat([t,n],1)
            if (n==eos).all(): break
        return t

print('All architectures loaded!')
```

```python
# ==== LOAD MODELS ====
from transformers import WhisperProcessor,WhisperForConditionalGeneration,NllbTokenizer,AutoModelForSeq2SeqLM
from gtts import gTTS
import tempfile

ASR_PATH='/kaggle/input/multilingual-asr-model/asr_model.pt'
NMT_PATH='/kaggle/input/multilingual-nmt-model/nmt_model.pt'

nllb_tok=NllbTokenizer.from_pretrained('facebook/nllb-200-distilled-600M')
VOCAB_SIZE=nllb_tok.vocab_size; PAD_ID=nllb_tok.pad_token_id; BOS_ID=nllb_tok.bos_token_id; EOS_ID=nllb_tok.eos_token_id

# ASR
asr_custom=None; use_custom_asr=os.path.exists(ASR_PATH)
if use_custom_asr:
    try:
        asr_custom=MicroConformerASR(vocab_size=5000).to(DEVICE)
        asr_custom.load_state_dict(torch.load(ASR_PATH,map_location=DEVICE)); asr_custom.eval()
        print('Custom ASR loaded!')
    except Exception as e: print(f'ASR fallback: {e}'); use_custom_asr=False
if not use_custom_asr:
    wh_proc=WhisperProcessor.from_pretrained('openai/whisper-base')
    wh_model=WhisperForConditionalGeneration.from_pretrained('openai/whisper-base').to(DEVICE)
    wh_model.eval(); print('Whisper loaded (fallback)')

# NMT
nmt_custom=None; use_custom_nmt=os.path.exists(NMT_PATH)
if use_custom_nmt:
    try:
        nmt_custom=MultilingualSeq2Seq(vocab_size=VOCAB_SIZE,pad_id=PAD_ID).to(DEVICE)
        nmt_custom.load_state_dict(torch.load(NMT_PATH,map_location=DEVICE)); nmt_custom.eval()
        print('Custom NMT loaded!')
    except Exception as e: print(f'NMT fallback: {e}'); use_custom_nmt=False
if not use_custom_nmt:
    nllb_model=AutoModelForSeq2SeqLM.from_pretrained('facebook/nllb-200-distilled-600M',dtype=torch.float16).to(DEVICE)
    nllb_model.eval(); print('NLLB-200 loaded (fallback)')

print('All models ready!')
```

```python
LANGUAGES={
    'Hindi':{'nllb':'hin_Deva','gtts':'hi','flag':'🇮🇳'},
    'Bengali':{'nllb':'ben_Beng','gtts':'bn','flag':'🇧🇩'},
    'Tamil':{'nllb':'tam_Taml','gtts':'ta','flag':'🇮🇳'},
    'Telugu':{'nllb':'tel_Telu','gtts':'te','flag':'🇮🇳'},
    'Marathi':{'nllb':'mar_Deva','gtts':'mr','flag':'🇮🇳'},
    'Gujarati':{'nllb':'guj_Gujr','gtts':'gu','flag':'🇮🇳'},
    'Kannada':{'nllb':'kan_Knda','gtts':'kn','flag':'🇮🇳'},
    'Malayalam':{'nllb':'mal_Mlym','gtts':'ml','flag':'🇮🇳'},
    'Punjabi':{'nllb':'pan_Guru','gtts':'pa','flag':'🇮🇳'},
    'Urdu':{'nllb':'urd_Arab','gtts':'ur','flag':'🇵🇰'},
    'Japanese':{'nllb':'jpn_Jpan','gtts':'ja','flag':'🇯🇵'},
    'Chinese':{'nllb':'zho_Hans','gtts':'zh-CN','flag':'🇨🇳'},
    'German':{'nllb':'deu_Latn','gtts':'de','flag':'🇩🇪'},
    'French':{'nllb':'fra_Latn','gtts':'fr','flag':'🇫🇷'},
    'Dutch':{'nllb':'nld_Latn','gtts':'nl','flag':'🇳🇱'},
    'Spanish':{'nllb':'spa_Latn','gtts':'es','flag':'🇪🇸'},
    'Italian':{'nllb':'ita_Latn','gtts':'it','flag':'🇮🇹'},
    'Portuguese':{'nllb':'por_Latn','gtts':'pt','flag':'🇵🇹'},
    'Russian':{'nllb':'rus_Cyrl','gtts':'ru','flag':'🇷🇺'},
    'Arabic':{'nllb':'arb_Arab','gtts':'ar','flag':'🇸🇦'},
    'Korean':{'nllb':'kor_Hang','gtts':'ko','flag':'🇰🇷'},
    'Turkish':{'nllb':'tur_Latn','gtts':'tr','flag':'🇹🇷'},
    'Polish':{'nllb':'pol_Latn','gtts':'pl','flag':'🇵🇱'},
    'Swedish':{'nllb':'swe_Latn','gtts':'sv','flag':'🇸🇪'},
    'Greek':{'nllb':'ell_Grek','gtts':'el','flag':'🇬🇷'},
    'Hebrew':{'nllb':'heb_Hebr','gtts':'iw','flag':'🇮🇱'},
    'Thai':{'nllb':'tha_Thai','gtts':'th','flag':'🇹🇭'},
    'Vietnamese':{'nllb':'vie_Latn','gtts':'vi','flag':'🇻🇳'},
}
LANG_NAMES=list(LANGUAGES.keys())

def do_transcribe(audio_path):
    if not audio_path: return 'No audio'
    try:
        wav,sr=torchaudio.load(audio_path)
        if sr!=16000: wav=torchaudio.functional.resample(wav,sr,16000)
        wav=wav.mean(0,keepdim=True)
        if use_custom_asr:
            toks=asr_custom.greedy_decode(wav.to(DEVICE),bos=BOS_ID,eos=EOS_ID)
            return nllb_tok.decode(toks[0].cpu(),skip_special_tokens=True)
        else:
            inp=wh_proc(wav.squeeze(0).numpy(),sampling_rate=16000,return_tensors='pt')
            ids=wh_model.generate(inp.input_features.to(DEVICE))
            return wh_proc.decode(ids[0],skip_special_tokens=True)
    except Exception as e: return f'Error: {e}'

def do_translate(text,tgt_nllb):
    if not text.strip(): return ''
    try:
        if use_custom_nmt:
            nllb_tok.src_lang='eng_Latn'
            ids=nllb_tok(text,return_tensors='pt',truncation=True,max_length=128)['input_ids'].to(DEVICE)
            out=nmt_custom.translate(ids,bos=BOS_ID,eos=EOS_ID)
            return nllb_tok.decode(out[0].cpu(),skip_special_tokens=True)
        else:
            nllb_tok.src_lang='eng_Latn'
            inp=nllb_tok(text,return_tensors='pt',truncation=True,max_length=128).to(DEVICE)
            tid=nllb_tok.convert_tokens_to_ids(tgt_nllb)
            out=nllb_model.generate(**inp,forced_bos_token_id=tid,max_new_tokens=200)
            return nllb_tok.decode(out[0],skip_special_tokens=True)
    except Exception as e: return f'[Error: {e}]'

def do_speak(text,gtts_code):
    if not text.strip(): return None
    try:
        tmp=tempfile.NamedTemporaryFile(suffix='.mp3',delete=False)
        gTTS(text=text,lang=gtts_code,slow=False).save(tmp.name)
        return tmp.name
    except: return None

print('Inference functions ready!')
```

```python
import gradio as gr

def translate_single(text,lang_name):
    info=LANGUAGES.get(lang_name,{})
    t=do_translate(text,info.get('nllb','hin_Deva'))
    a=do_speak(t,info.get('gtts','hi'))
    return t,a

def translate_all(text):
    return {f"{v['flag']} {k}": do_translate(text,v['nllb']) for k,v in LANGUAGES.items()}

def pipeline(audio,lang):
    txt=do_transcribe(audio)
    t,a=translate_single(txt,lang)
    return txt,t,a

with gr.Blocks(title='Multilingual AI Translator',theme=gr.themes.Soft()) as demo:
    gr.Markdown('# 🌐 Multilingual AI Language Translator\n49 languages · Custom Conformer ASR + Seq2Seq NMT + FastSpeech2 TTS')
    with gr.Tabs():
        with gr.Tab('🎙️ Live Recording'):
            with gr.Row():
                with gr.Column():
                    audio_in=gr.Audio(sources=['microphone','upload'],type='filepath',label='Audio Input')
                    lang_sel=gr.Dropdown(choices=LANG_NAMES,value='Hindi',label='Target Language')
                    btn=gr.Button('Transcribe & Translate',variant='primary')
                with gr.Column():
                    txt_out=gr.Textbox(label='Transcribed Text',lines=3)
                    tr_out=gr.Textbox(label='Translation',lines=3)
                    aud_out=gr.Audio(label='Voice Output',type='filepath')
            btn.click(fn=pipeline,inputs=[audio_in,lang_sel],outputs=[txt_out,tr_out,aud_out])
        with gr.Tab('📝 Text Input'):
            with gr.Row():
                with gr.Column():
                    ti=gr.Textbox(label='English Text',lines=4,placeholder='Type here...')
                    ld=gr.Dropdown(choices=LANG_NAMES,value='Hindi',label='Language')
                    tb=gr.Button('Translate & Speak',variant='primary')
                with gr.Column():
                    to=gr.Textbox(label='Translation',lines=4)
                    ao=gr.Audio(label='Voice',type='filepath')
            tb.click(fn=translate_single,inputs=[ti,ld],outputs=[to,ao])
        with gr.Tab('🌍 All Languages'):
            at=gr.Textbox(label='English Text',lines=3,placeholder='Translate to all...')
            ab=gr.Button('Translate to All 28 Languages',variant='primary')
            ao2=gr.JSON(label='Translations')
            ab.click(fn=translate_all,inputs=[at],outputs=[ao2])

print('Launching Gradio with share=True ...')
demo.launch(share=True,server_port=7860,server_name='0.0.0.0')
```
