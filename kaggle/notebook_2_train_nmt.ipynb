```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import math
from typing import Optional

class RoPE(nn.Module):
    def __init__(self,dh,ml=2048):
        super().__init__()
        inv=1.0/(10000**(torch.arange(0,dh,2).float()/dh))
        self.register_buffer('inv',inv)
    def _rh(self,x): x1,x2=x[...,::2],x[...,1::2]; return torch.stack([-x2,x1],-1).flatten(-2)
    def forward(self,x):
        B,H,T,D=x.shape; t=torch.arange(T,device=x.device).float()
        f=torch.einsum('i,j->ij',t,self.inv); e=torch.cat([f,f],-1).unsqueeze(0).unsqueeze(0)
        return x*e.cos()+self._rh(x)*e.sin()

class MHARoPE(nn.Module):
    def __init__(self,d,h,p=0.1,use_rope=True):
        super().__init__()
        self.dh=d//h; self.h=h; self.rp=use_rope
        self.q=nn.Linear(d,d); self.k=nn.Linear(d,d); self.v=nn.Linear(d,d)
        self.o=nn.Linear(d,d); self.drop=nn.Dropout(p); self.sc=self.dh**-0.5
        if use_rope: self.rope=RoPE(self.dh)
    def forward(self,q,k,v,mask=None):
        B=q.size(0)
        def sp(t): return t.view(B,-1,self.h,self.dh).transpose(1,2)
        Q,K,V=sp(self.q(q)),sp(self.k(k)),sp(self.v(v))
        if self.rp: Q=self.rope(Q); K=self.rope(K)
        s=(Q@K.transpose(-2,-1))*self.sc
        if mask is not None: s=s.masked_fill(mask,float('-inf'))
        return self.drop(self.o((self.drop(F.softmax(s,-1))@V).transpose(1,2).reshape(B,-1,self.h*self.dh)))

class PnFFN(nn.Module):
    def __init__(self,d,f,p=0.1):
        super().__init__()
        self.n=nn.LayerNorm(d); self.f1=nn.Linear(d,f); self.f2=nn.Linear(f,d); self.drop=nn.Dropout(p); self.act=nn.GELU()
    def forward(self,x): return x+self.drop(self.f2(self.drop(self.act(self.f1(self.n(x))))))

class EncBlk(nn.Module):
    def __init__(self,d=512,h=8,f=2048,p=0.1):
        super().__init__()
        self.n=nn.LayerNorm(d); self.sa=MHARoPE(d,h,p,True); self.ffn=PnFFN(d,f,p)
    def forward(self,x,mask=None):
        xn=self.n(x); return self.ffn(x+self.sa(xn,xn,xn,mask))

class DecBlk(nn.Module):
    def __init__(self,d=512,h=8,f=2048,p=0.1):
        super().__init__()
        self.n1=nn.LayerNorm(d); self.sa=MHARoPE(d,h,p,True)
        self.n2=nn.LayerNorm(d); self.ca=MHARoPE(d,h,p,False); self.ffn=PnFFN(d,f,p)
    def forward(self,t,m,tm=None,mm=None):
        xn=self.n1(t); t=t+self.sa(xn,xn,xn,tm)
        xn=self.n2(t); t=t+self.ca(xn,m,m,mm); return self.ffn(t)

class LabelSmoothing(nn.Module):
    def __init__(self,vs,s=0.1,ig=0):
        super().__init__()
        self.s=s; self.vs=vs; self.ig=ig
    def forward(self,logits,targets):
        lp=F.log_softmax(logits,-1); BT,V=lp.size()
        with torch.no_grad():
            sm=torch.full_like(lp,self.s/(V-2))
            sm.scatter_(1,targets.unsqueeze(1).clamp(min=0),1.0-self.s)
            sm[:,self.ig]=0; mask=(targets==self.ig); sm[mask]=0
        loss=-(sm*lp).sum(-1)
        return loss[~mask].mean()

class MultilingualSeq2Seq(nn.Module):
    def __init__(self,vocab_size=32000,d_model=512,n_enc=6,n_dec=6,n_heads=8,ffn_dim=2048,dropout=0.1,max_len=512,pad_id=0):
        super().__init__()
        self.pad=pad_id; self.d=d_model
        self.em=nn.Embedding(vocab_size,d_model,padding_idx=pad_id)
        self.enc=nn.ModuleList([EncBlk(d_model,n_heads,ffn_dim,dropout) for _ in range(n_enc)])
        self.dec=nn.ModuleList([DecBlk(d_model,n_heads,ffn_dim,dropout) for _ in range(n_dec)])
        self.en=nn.LayerNorm(d_model); self.dn=nn.LayerNorm(d_model)
        self.lm=nn.Linear(d_model,vocab_size,bias=False); self.lm.weight=self.em.weight
        self.loss_fn=LabelSmoothing(vocab_size,0.1,pad_id); self.drop=nn.Dropout(dropout)
    def _cm(self,T,dev): return torch.triu(torch.ones(T,T,device=dev),1).bool().unsqueeze(0).unsqueeze(0)
    def encode(self,src):
        mask=(src==self.pad).unsqueeze(1).unsqueeze(2)
        x=self.drop(self.em(src))*math.sqrt(self.d)
        for b in self.enc: x=b(x,mask)
        return self.en(x)
    def decode(self,tgt,mem):
        T=tgt.size(1); tm=self._cm(T,tgt.device)
        x=self.drop(self.em(tgt))*math.sqrt(self.d)
        for b in self.dec: x=b(x,mem,tm)
        return self.dn(x)
    def forward(self,src,tgt):
        m=self.encode(src); h=self.decode(tgt[:,:-1],m); return self.lm(h)
    def compute_loss(self,src,tgt):
        logits=self(src,tgt); targets=tgt[:,1:].contiguous()
        return self.loss_fn(logits.reshape(-1,logits.size(-1)),targets.reshape(-1))
    @torch.no_grad()
    def translate(self,src,max_len=200,bos=1,eos=2):
        self.eval(); m=self.encode(src); B=src.size(0)
        t=torch.full((B,1),bos,dtype=torch.long,device=src.device)
        for _ in range(max_len):
            h=self.decode(t,m); n=self.lm(h[:,-1,:]).argmax(-1,keepdim=True)
            t=torch.cat([t,n],1)
            if (n==eos).all(): break
        return t

m=MultilingualSeq2Seq(vocab_size=32000)
total=sum(p.numel() for p in m.parameters())
print(f'MultilingualSeq2Seq: {total:,} ({total/1e6:.1f}M) parameters')
del m; print('Architecture OK!')
```

```python
from transformers import NllbTokenizer
tokenizer=NllbTokenizer.from_pretrained('facebook/nllb-200-distilled-600M')
VOCAB_SIZE=tokenizer.vocab_size; PAD_ID=tokenizer.pad_token_id
BOS_ID=tokenizer.bos_token_id; EOS_ID=tokenizer.eos_token_id
print(f'NLLB vocab={VOCAB_SIZE} PAD={PAD_ID} BOS={BOS_ID} EOS={EOS_ID}')
LANG_PAIRS=[('eng_Latn','hin_Deva'),('eng_Latn','ben_Beng'),('eng_Latn','tam_Taml'),
            ('eng_Latn','tel_Telu'),('eng_Latn','mar_Deva'),('eng_Latn','guj_Gujr'),
            ('eng_Latn','jpn_Jpan'),('eng_Latn','zho_Hans'),('eng_Latn','deu_Latn'),
            ('eng_Latn','fra_Latn'),('eng_Latn','rus_Cyrl'),('eng_Latn','arb_Arab'),('eng_Latn','kor_Hang')]
print(f'{len(LANG_PAIRS)} language pairs configured')
```

```python
from datasets import load_dataset
from torch.utils.data import Dataset,DataLoader,ConcatDataset
import torch

MAX_LEN=128

class NMTDataset(Dataset):
    def __init__(self,examples,tok,sl,tl,ml=128):
        self.ex=examples; self.tok=tok; self.sl=sl; self.tl=tl; self.ml=ml
    def __len__(self): return len(self.ex)
    def __getitem__(self,idx):
        pair=self.ex[idx]
        tr=pair.get('translation',{})
        sl_code=self.sl.split('_')[0]; tl_code=self.tl.split('_')[0]
        src_txt=tr.get(sl_code,tr.get('en',''))
        tgt_txt=tr.get(tl_code,tr.get(sl_code,''))
        self.tok.src_lang=self.sl
        src=self.tok(src_txt,max_length=self.ml,truncation=True,padding='max_length',return_tensors='pt')
        self.tok.src_lang=self.tl
        tgt=self.tok(tgt_txt,max_length=self.ml,truncation=True,padding='max_length',return_tensors='pt')
        return src['input_ids'].squeeze(0),tgt['input_ids'].squeeze(0)

all_ds=[]
for sl,tl in LANG_PAIRS[:6]:
    lc=tl.split('_')[0]
    try:
        ds=load_dataset('ai4bharat/samanantar',f'en-{lc}',split='train[:8000]',trust_remote_code=True)
        all_ds.append(NMTDataset(ds,tokenizer,sl,tl,MAX_LEN)); print(f'Samanantar en-{lc}: {len(ds)}')
    except:
        try:
            ds=load_dataset('Helsinki-NLP/opus-100',f'en-{lc}',split='train[:5000]',trust_remote_code=True)
            all_ds.append(NMTDataset(ds,tokenizer,sl,tl,MAX_LEN)); print(f'OPUS en-{lc}: {len(ds)}')
        except Exception as e: print(f'Skip {lc}: {e}')

combined=ConcatDataset(all_ds)
train_loader=DataLoader(combined,batch_size=16,shuffle=True,num_workers=2,pin_memory=True,
                        collate_fn=lambda b:(torch.stack([x[0] for x in b]),torch.stack([x[1] for x in b])))
print(f'Total: {len(combined)} samples | {len(train_loader)} batches/epoch')
```

```python
import time,json
DEVICE=torch.device('cuda' if torch.cuda.is_available() else 'cpu')
NUM_EPOCHS=25; LR=1e-4
SAVE_PATH='/kaggle/working/nmt_model.pt'

model=MultilingualSeq2Seq(vocab_size=VOCAB_SIZE,pad_id=PAD_ID).to(DEVICE)
optimizer=torch.optim.AdamW(model.parameters(),lr=LR,weight_decay=1e-5)
scheduler=torch.optim.lr_scheduler.OneCycleLR(optimizer,max_lr=LR,steps_per_epoch=len(train_loader),epochs=NUM_EPOCHS)
scaler=torch.cuda.amp.GradScaler(enabled=torch.cuda.is_available())
print(f'{sum(p.numel() for p in model.parameters())/1e6:.1f}M params | device: {DEVICE}')

best=float('inf'); hist=[]
for epoch in range(1,NUM_EPOCHS+1):
    model.train(); el=0.0; t0=time.time()
    for i,(src,tgt) in enumerate(train_loader):
        src,tgt=src.to(DEVICE),tgt.to(DEVICE); optimizer.zero_grad()
        with torch.cuda.amp.autocast(enabled=torch.cuda.is_available()):
            loss=model.compute_loss(src,tgt)
        if torch.isnan(loss): continue
        scaler.scale(loss).backward(); scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(),1.0)
        scaler.step(optimizer); scaler.update(); scheduler.step(); el+=loss.item()
        if i%100==0: print(f'  Ep{epoch:02d} [{i}/{len(train_loader)}] loss={loss.item():.4f}')
    avg=el/max(1,len(train_loader)); hist.append({'epoch':epoch,'loss':avg})
    print(f'Epoch {epoch:02d} avg={avg:.4f} | {time.time()-t0:.0f}s')
    torch.save({'epoch':epoch,'state':model.state_dict(),'loss':avg},SAVE_PATH.replace('.pt','_ckpt.pt'))
    if avg<best: best=avg; torch.save(model.state_dict(),SAVE_PATH); print(f'  Best! {best:.4f}')

with open('/kaggle/working/nmt_history.json','w') as f: json.dump(hist,f)
print(f'Done! Best={best:.4f} | {SAVE_PATH}')
```
