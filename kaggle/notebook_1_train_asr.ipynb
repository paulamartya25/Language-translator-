```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torchaudio
import math
from typing import Optional

class LogMelSpec(nn.Module):
    def __init__(self, sr=16000, n_fft=512, hop=160, n_mels=80):
        super().__init__()
        self.ms = torchaudio.transforms.MelSpectrogram(sample_rate=sr,n_fft=n_fft,hop_length=hop,n_mels=n_mels)
        self.db = torchaudio.transforms.AmplitudeToDB(top_db=80)
    def forward(self,w):
        m=self.db(self.ms(w)); return (m-m.mean(-1,keepdim=True))/(m.std(-1,keepdim=True)+1e-6)

class ConvSub(nn.Module):
    def __init__(self,in_c,d):
        super().__init__()
        self.c=nn.Sequential(nn.Conv2d(1,256,3,2,1),nn.ReLU(),nn.Conv2d(256,256,3,2,1),nn.ReLU())
        self.p=nn.Linear(256*((in_c+1)//4),d)
    def forward(self,x):
        x=self.c(x.unsqueeze(1)); B,C,F,T=x.size()
        return self.p(x.permute(0,3,1,2).reshape(B,T,C*F))

class Swish(nn.Module):
    def forward(self,x): return x*torch.sigmoid(x)

class ConvMod(nn.Module):
    def __init__(self,d,k=31):
        super().__init__()
        self.n=nn.LayerNorm(d); self.p1=nn.Conv1d(d,2*d,1); self.glu=nn.GLU(1)
        self.dw=nn.Conv1d(d,d,k,padding=(k-1)//2,groups=d); self.bn=nn.BatchNorm1d(d)
        self.sw=Swish(); self.p2=nn.Conv1d(d,d,1); self.drop=nn.Dropout(0.1)
    def forward(self,x):
        r=x; x=self.n(x).transpose(1,2); x=self.sw(self.bn(self.dw(self.glu(self.p1(x)))))
        return self.drop(self.p2(x)).transpose(1,2)+r

class FFMod(nn.Module):
    def __init__(self,d,e=4,p=0.1):
        super().__init__()
        self.n=nn.LayerNorm(d); self.f1=nn.Linear(d,d*e); self.a=Swish()
        self.d1=nn.Dropout(p); self.f2=nn.Linear(d*e,d); self.d2=nn.Dropout(p)
    def forward(self,x): return x+0.5*self.d2(self.f2(self.d1(self.a(self.f1(self.n(x))))))

class MHSA(nn.Module):
    def __init__(self,d,h,p=0.1):
        super().__init__()
        self.dh=d//h; self.h=h; self.n=nn.LayerNorm(d)
        self.q=nn.Linear(d,d); self.k=nn.Linear(d,d); self.v=nn.Linear(d,d)
        self.o=nn.Linear(d,d); self.drop=nn.Dropout(p); self.sc=self.dh**-0.5
    def forward(self,x,mask=None):
        r=x; x=self.n(x); B,T,_=x.shape
        def sp(t): return t.view(B,-1,self.h,self.dh).transpose(1,2)
        Q,K,V=sp(self.q(x)),sp(self.k(x)),sp(self.v(x))
        s=(Q@K.transpose(-2,-1))*self.sc
        if mask is not None: s=s.masked_fill(mask.unsqueeze(1).unsqueeze(2),float('-inf'))
        return self.drop(self.o((self.drop(F.softmax(s,-1))@V).transpose(1,2).reshape(B,T,-1)))+r

class ConfBlk(nn.Module):
    def __init__(self,d=256,h=4,k=31,p=0.1):
        super().__init__()
        self.ff1=FFMod(d,p=p); self.sa=MHSA(d,h,p); self.cv=ConvMod(d,k); self.ff2=FFMod(d,p=p); self.n=nn.LayerNorm(d)
    def forward(self,x,mask=None): return self.n(self.ff2(self.cv(self.sa(self.ff1(x),mask))))

class MHA(nn.Module):
    def __init__(self,d,h,p=0.1):
        super().__init__()
        self.dh=d//h; self.h=h; self.q=nn.Linear(d,d); self.k=nn.Linear(d,d); self.v=nn.Linear(d,d)
        self.o=nn.Linear(d,d); self.drop=nn.Dropout(p); self.sc=self.dh**-0.5
    def forward(self,q,k,v,mask=None):
        B=q.size(0)
        def sp(t): return t.view(B,-1,self.h,self.dh).transpose(1,2)
        Q,K,V=sp(self.q(q)),sp(self.k(k)),sp(self.v(v)); s=(Q@K.transpose(-2,-1))*self.sc
        if mask is not None: s=s.masked_fill(mask,float('-inf'))
        return self.o((self.drop(F.softmax(s,-1))@V).transpose(1,2).reshape(B,-1,self.h*self.dh))

class DecBlk(nn.Module):
    def __init__(self,d=256,h=4,f=1024,p=0.1):
        super().__init__()
        self.n1=nn.LayerNorm(d); self.sa=MHA(d,h,p)
        self.n2=nn.LayerNorm(d); self.ca=MHA(d,h,p)
        self.n3=nn.LayerNorm(d); self.ff=nn.Sequential(nn.Linear(d,f),nn.ReLU(),nn.Dropout(p),nn.Linear(f,d),nn.Dropout(p))
    def forward(self,t,m,tm=None,mm=None):
        x=t; x=x+self.sa(self.n1(x),self.n1(x),self.n1(x),tm)
        x=x+self.ca(self.n2(x),m,m,mm); return x+self.ff(self.n3(x))

class TDec(nn.Module):
    def __init__(self,vs,d=256,L=6,h=4,f=1024,p=0.1,ml=512):
        super().__init__()
        self.em=nn.Embedding(vs,d,padding_idx=0); self.pe=nn.Embedding(ml,d)
        self.drop=nn.Dropout(p); self.blocks=nn.ModuleList([DecBlk(d,h,f,p) for _ in range(L)]); self.n=nn.LayerNorm(d)
    def forward(self,t,m,tm=None,mm=None):
        B,T=t.size(); pos=torch.arange(T,device=t.device).unsqueeze(0)
        x=self.drop(self.em(t)+self.pe(pos))
        for b in self.blocks: x=b(x,m,tm,mm)
        return self.n(x)

class MicroConformerASR(nn.Module):
    def __init__(self,vocab_size=5000,n_mels=80,d_model=256,enc_layers=6,dec_layers=6,n_heads=4,kernel_size=31,ffn_dim=1024,dropout=0.1,max_len=512,ctc_weight=0.3):
        super().__init__()
        self.ctc_w=ctc_weight
        self.fe=LogMelSpec(n_mels=n_mels); self.sub=ConvSub(n_mels,d_model)
        self.drop=nn.Dropout(dropout)
        self.enc=nn.ModuleList([ConfBlk(d_model,n_heads,kernel_size,dropout) for _ in range(enc_layers)])
        self.enc_n=nn.LayerNorm(d_model)
        self.dec=TDec(vocab_size,d_model,dec_layers,n_heads,ffn_dim,dropout,max_len)
        self.ctc=nn.Linear(d_model,vocab_size)
        self.lm=nn.Linear(d_model,vocab_size); self.lm.weight=self.dec.em.weight
    def encode(self,w,mask=None):
        x=self.sub(self.fe(w))
        for b in self.enc: x=b(x,mask)
        return self.enc_n(x)
    def forward(self,w,t,mask=None):
        m=self.encode(w,mask)
        return F.log_softmax(self.ctc(m),-1),self.lm(self.dec(t[:,:-1],m))
    def compute_loss(self,w,t,enc_len=None,tgt_len=None):
        B=w.size(0); ctc_l,ce_l=self(w,t)
        T=ctc_l.size(1)
        if enc_len is None: enc_len=torch.full((B,),T,dtype=torch.long,device=w.device)
        if tgt_len is None: tgt_len=torch.full((B,),t.size(1)-1,dtype=torch.long,device=w.device)
        ctc_loss=F.ctc_loss(ctc_l.transpose(0,1),t[:,1:],enc_len,tgt_len,blank=0,zero_infinity=True)
        ce_loss=F.cross_entropy(ce_l.reshape(-1,ce_l.size(-1)),t[:,1:].reshape(-1),ignore_index=0)
        return self.ctc_w*ctc_loss+(1-self.ctc_w)*ce_loss
    @torch.no_grad()
    def greedy_decode(self,w,max_len=200,bos=1,eos=2):
        self.eval(); m=self.encode(w); B=w.size(0)
        toks=torch.full((B,1),bos,dtype=torch.long,device=w.device)
        for _ in range(max_len):
            n=self.lm(self.dec(toks,m))[:,-1,:].argmax(-1,keepdim=True)
            toks=torch.cat([toks,n],1)
            if (n==eos).all(): break
        return toks

model_test=MicroConformerASR(vocab_size=5000)
total=sum(p.numel() for p in model_test.parameters())
print(f'MicroConformerASR: {total:,} ({total/1e6:.1f}M) parameters')
del model_test
print('Architecture OK!')
```

```python
from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained('facebook/mms-300m', use_fast=True)
VOCAB_SIZE = tokenizer.vocab_size
BOS_ID = tokenizer.bos_token_id or 1
EOS_ID = tokenizer.eos_token_id or 2
PAD_ID = tokenizer.pad_token_id or 0
print(f'Vocab: {VOCAB_SIZE} | BOS: {BOS_ID} | EOS: {EOS_ID}')
```

```python
from datasets import load_dataset, concatenate_datasets
from torch.utils.data import Dataset, DataLoader
import torch

LANGUAGES = ['hi_in','bn_in','ta_in','te_in','mr_in','gu_in','kn_in','ml_in']
datasets_list = []
for lang in LANGUAGES:
    try:
        ds = load_dataset('google/fleurs', lang, split='train', trust_remote_code=True)
        datasets_list.append(ds); print(f'  FLEURS {lang}: {len(ds)} samples')
    except Exception as e: print(f'  Skip {lang}: {e}')
try:
    en_ds = load_dataset('mozilla-foundation/common_voice_13_0','en',split='train[:5000]',trust_remote_code=True)
    datasets_list.append(en_ds); print(f'  CommonVoice EN: {len(en_ds)} samples')
except Exception as e: print(f'  Skip EN: {e}')
print(f'Total datasets: {len(datasets_list)}')

class ASRDataset(Dataset):
    def __init__(self, hf_ds, tok, max_audio=160000, max_text=200):
        self.data=hf_ds; self.tok=tok; self.ma=max_audio; self.mt=max_text
    def __len__(self): return len(self.data)
    def __getitem__(self,idx):
        item=self.data[idx]; arr=item['audio']['array']; sr=item['audio']['sampling_rate']
        if sr!=16000:
            import torchaudio
            wav=torch.tensor(arr,dtype=torch.float32).unsqueeze(0)
            wav=torchaudio.functional.resample(wav,sr,16000).squeeze(0).numpy()
        else: wav=arr
        waveform=torch.tensor(wav[:self.ma],dtype=torch.float32)
        text=item.get('transcription',item.get('sentence',''))
        toks=self.tok.encode(text,add_special_tokens=True,max_length=self.mt,truncation=True)
        return waveform, torch.tensor(toks,dtype=torch.long)

def collate(batch):
    wavs,toks=zip(*batch)
    mw=max(w.size(0) for w in wavs); mt=max(t.size(0) for t in toks)
    pw=torch.zeros(len(wavs),mw); pt=torch.zeros(len(toks),mt,dtype=torch.long)
    for i,(w,t) in enumerate(zip(wavs,toks)):
        pw[i,:w.size(0)]=w; pt[i,:t.size(0)]=t
    return pw,pt

combined=concatenate_datasets(datasets_list)
asr_ds=ASRDataset(combined,tokenizer)
train_loader=DataLoader(asr_ds,batch_size=8,shuffle=True,collate_fn=collate,num_workers=2,pin_memory=True)
print(f'DataLoader ready: {len(asr_ds)} samples, {len(train_loader)} batches/epoch')
```

```python
import time,json
DEVICE=torch.device('cuda' if torch.cuda.is_available() else 'cpu')
NUM_EPOCHS=20; LR=3e-4; GRAD_CLIP=5.0
SAVE_PATH='/kaggle/working/asr_model.pt'; CKPT='/kaggle/working/asr_ckpt.pt'

model=MicroConformerASR(vocab_size=VOCAB_SIZE,ctc_weight=0.3).to(DEVICE)
optimizer=torch.optim.AdamW(model.parameters(),lr=LR,weight_decay=1e-6)
scheduler=torch.optim.lr_scheduler.CosineAnnealingLR(optimizer,T_max=NUM_EPOCHS)
scaler=torch.cuda.amp.GradScaler(enabled=torch.cuda.is_available())
print(f'Training {sum(p.numel() for p in model.parameters())/1e6:.1f}M params on {DEVICE}')

best_loss=float('inf'); history=[]
for epoch in range(1,NUM_EPOCHS+1):
    model.train(); el=0.0; t0=time.time()
    for i,(wav,tgt) in enumerate(train_loader):
        wav,tgt=wav.to(DEVICE),tgt.to(DEVICE); optimizer.zero_grad()
        with torch.cuda.amp.autocast(enabled=torch.cuda.is_available()):
            loss=model.compute_loss(wav,tgt)
        if torch.isnan(loss): continue
        scaler.scale(loss).backward()
        scaler.unscale_(optimizer); torch.nn.utils.clip_grad_norm_(model.parameters(),GRAD_CLIP)
        scaler.step(optimizer); scaler.update(); el+=loss.item()
        if i%50==0: print(f'  Ep{epoch:02d} [{i}/{len(train_loader)}] loss={loss.item():.4f} lr={optimizer.param_groups[0]["lr"]:.2e}')
    scheduler.step(); avg=el/max(1,len(train_loader))
    history.append({'epoch':epoch,'loss':avg})
    print(f'Epoch {epoch:02d} | avg_loss={avg:.4f} | {time.time()-t0:.0f}s')
    torch.save({'epoch':epoch,'state':model.state_dict(),'loss':avg},CKPT)
    if avg<best_loss: best_loss=avg; torch.save(model.state_dict(),SAVE_PATH); print(f'  Best saved! {best_loss:.4f}')

with open('/kaggle/working/asr_history.json','w') as f: json.dump(history,f,indent=2)
print(f'Done! Best loss: {best_loss:.4f} | Saved: {SAVE_PATH}')
```
